# 4.6.2 List comprehensions

"Lijstmanipulaties on steroids" en daarom een grote hoofding.

De beste en meest gebruikte manier om berekeningen met een lijst in Python te doen met als algemeen formaat:

	nieuweLijst = [expressie for element in lijst if voorwaarde]

De eenvoudige variant is als volgt
### nieuweLijst = [f(element) for element in invoerlijst]

“Maak een nieuwe lijst door een functie f toe te passen op elk element in de invoerlijst”.

- [ ... for ... in ... ] zijn verplicht. Daar mag je niet van afwijken.
- *invoerlijst* moet je een bestaande lijst voor opgeven: [1,2,3], range(5,9), ... of een variabele met een lijst in.
- *element* is een naam die je zelf mag kiezen, zolang je maar consequent bent en dus 2x dezelfde naam gebruikt. Kan dus *i* of *getal* of *naam*, ... zijn
- *f(...)* is een functie die je loslaat op elk element


```python
def fahrenheit_to_celsius(fahrenheit):
    return (fahrenheit - 32) * 5 / 9

temp_in_f = [35, 40, 28, 52, 15]
temp_in_c = [fahrenheit_to_celsius(temp) for temp in temp_in_f]
print(temp_in_c)
```

```python
kwadraten = [i*i for i in [1,2,3,4] ]
print(kwadraten)
```

```python
xs = [1,2,3,4]
dubbeleXS = [x*2 for x in xs]
print(dubbeleXS)
```

In de uitgebreidere variant komt er een if bij. Die legt een voorwaarde op aan de elementen die geselecteerd worden uit de lijst, bijvoorbeeld

   `graden = [graad(student) for student in iiw if geslaagd(student)]`


```python
xs = range(20)
halveXS = [x//2 for x in xs if x%2 == 0]
print(halveXS)

import math
wortels = [math.sqrt(i) for i in xs if i >= 0]
print(wortels)
```

```python
def aantal_delers(getal):
    return len([x for x in range(1, getal+1) if getal % x == 0])

print(aantal_delers(2))
print(aantal_delers(3))
print(aantal_delers(14))
print(aantal_delers(16))
print(aantal_delers(17))

primes = [getal for getal in range(101) if aantal_delers(getal)== 2]
print(primes)
```

Wanneer het niet de bedoeling is om terug een lijst uit te komen en je bijvoorbeeld alle elementen wil afdrukken, dan moet je een andere structuur gebruiken die eigenlijk zegt "**Doe iets voor elk element van een lijst.**" Die structuur is de **for**:

Gebruik hiervoor onderstaand geraamte:

```python
for *element* in *lijst*:
   # doe iets met elk element bv.
   doeIets(*element*)
```

```python
## voorbeelden met 'echte' lijsten
def toon_lijst(lijst):
    for i in lijst:
        print(i)

toon_lijst("hallo")  # toont elke letter apart
```

Met een tuple werkt dat hetzelfde, maar je kan het tuple nu ook direct uitpakken.

```python
# 2e voorbeeld met echte lijst
# invoer = lijst van tuples met als eerste element "+", "-", "*" of "/" en tweede element.
# uitvoer = alle bewerkingen cumulatief uitgevoerd

def doe_bewerkingen(tuples):
    waarde = 0
    for bewerking, getal in tuples:
        if bewerking == "+":
            waarde = waarde + getal
        elif bewerking == "-":
            waarde = waarde - getal
        elif bewerking == "*":
            waarde = waarde * getal
        elif bewerking == "/":
            waarde = waarde / getal
    return waarde

print(doe_bewerkingen([("+",4),("-",8),("*",6),("/",2),("+",10),("*",5),("-",20)]))
```

## 4.7.2 enumerate

Dikwijls wil je tijdens het doorlopen van een for-lus een teller bijhouden van het hoeveelste element je aan het behandelen bent. Je kan daar manueel een variabele voor bijhouden, maar de enumerate doet dat voor jou. Let wel op: die telt vanaf 0.

Een eenvoudig voorbeeld:

```python
metingen = [12.4, 14.1, 13.8, 15.2]

# enumerate() geeft (0, 12.4), (1, 14.1), ... terug
for index, waarde in enumerate(metingen):
    print(f"Meting {index + 1}: {waarde} °C")
```

Via die index kan je een lijst ook vrij gemakkelijk opsplitsen in groepjes van 4. Elementen 0, 1, 2, 3 komen in de eerste lijst, elementen 4, 5,6,7 in de tweede, ... Telkens wanneer de index deelbaar is door vier begin je aan een nieuwe lijst.


```python
# voorbeeld voor enumerate
def per_4(lijst):
    huidige_lijst = []
    lijsten = []

    for index, element in enumerate(lijst):
        if index % 4 == 0 and index > 0:
            lijsten.append(huidige_lijst)
            huidige_lijst = []

        huidige_lijst.append(element)

    if len(huidige_lijst) > 0:
        lijsten.append(huidige_lijst)

    return lijsten

print(per_4(list(range(1,23))))
print(per_4("Filmproducent"))
print(per_4("Figurenjacht"))
```

```python
# oef 11
def luide_periodes(metingen):
    start_indices = []
    in_te_luide_periode = False

    for index, meting in enumerate(metingen):
        if meting > 95:
            if not in_te_luide_periode:
                start_indices.append(index)
                in_te_luide_periode = True
        else:
            in_te_luide_periode = False

    return start_indices


# Test met het eerdere voorbeeld:
# Indices met waarden > 95 zijn: 3 (waarde 96) en 8 (waarde 98)
geluidsdata = [89, 90, 93, 96, 97, 96, 94, 93, 98, 96]
print(luide_periodes(geluidsdata))
```

12. In het voorbeeld van `per_4` zetten we de eerste 4 elementen samen, dan de volgende vier elementen. Maak nu een variant zodat je een (grote) groep in vier opsplitst en om beurten een elementen in de vier deelgroepen steekt. Dus 0, 4, 8, ... in lijst1 steken, 1, 5, 9, ... in lijst 2 enzoverder. Dit principe noemen we in de informatica *round robin*.

```python
# oef. 12
def verdeel_over_4(lijst):
    # Begin met vier lege deellijsten
    resultaat = [[], [], [], []]

    for index, element in enumerate(lijst):
        doellijst_index = index % 4
        resultaat[doellijst_index].append(element)

    return resultaat

print(verdeel_over_4(list(range(1, 13))))
```

13. Maak een variant van de vorige opgave met een extra parameter die aangeeft in hoeveel groepen je moet verdelen.

```python
def verdeel_over_groepen(lijst, aantal_groepen):
    resultaat = [[] for i in range(aantal_groepen)]

    for index, element in enumerate(lijst):
        doellijst_index = index % aantal_groepen
        resultaat[doellijst_index].append(element)

    return resultaat


print(verdeel_over_groepen(list(range(1, 13)), 3))

print(verdeel_over_groepen(list(range(1, 13)), 5))
```
