# 3.5	Optionele en benoemde parameters

Wanneer je parameters hebt die een extra vormen en dus niet noodzakelijk zijn, kan je er optionele parameters van maken door een standaard-waarde achter de parameter te zetten.

```python
# voorbeeld uit de cursus
def hallo(naam, aantal_extra = 1, sierteken = '*'):
    aantal_sterren = len(naam) + (aantal_extra*2) + 8
    veel_sterren = sierteken * aantal_sterren
    extra_sterren = sierteken * aantal_extra
    print(veel_sterren)
    print(f"{extra_sterren} Hallo {naam} {extra_sterren}")
    print(veel_sterren)
```

```python
hallo("Dag allemaal", sierteken='#')
```

```python
hallo("PIT = Python als Ingenieurstool", sierteken="!", aantal_extra=3)
```

# 3.6 Functies als variable

Je kan ook functies als parameter meegeven aan een andere functies.

```python
# voorbeelden uit de cursus
def kleinste_waarde(f,g,x):
    if f(x) < g(x):
        return f(x)
    else:
        return g(x)

def f_positief_or_default(f, x, default ):
    if f(x) > 0:
         return f(x)
    else:
         return default
```

```python
# voorbeelden toegepast
def verdubbel(x):
    return x*2
def kwadrateer(x):
    return x**2

print("Kleinste waarde van verdubbel(1.5) en kwadrateer(1.5)")
print(kleinste_waarde(verdubbel, kwadrateer, 1.5))
print("Kleinste waarde van verdubbel(15) en kwadrateer(15)")
print(kleinste_waarde(verdubbel, kwadrateer, 15))

print("f_positief_or_default van verdubbel -1.5")
print(f_positief_or_default(verdubbel, -1.5, 0))
```

# 3.7 Functies in functies

Hieronder vind je een functie is_lokaal_minimum die drie parameters neemt:
- een functie f
- een getal x
- en een (klein) getal delta

De functie gaat na of f een lokaal minium in f heeft door te kijken of de functie daalt tussen x-delta en x (= stijgt tussen x en x-delta) en tussen x en x+delta.

```python
def is_stijgend(f, x, y):
    return f(x) < f(y)

def is_lokaal_minimum(f, x, delta):
    return is_stijgend(f, x, x-delta) and is_stijgend(f, x, x+delta)

def functie(x):
    if x < 0:
        return -x
    else:
        return x
    
ja = is_lokaal_minimum(functie, 0, 0.1)
nee= is_lokaal_minimum(functie, 1, 0.1)
print(f"{ja} en {nee}")
```

Stel nu dat we voor een functie met twéé parameters iets soortgelijks willen doen. Concreet willen we weten of er in het punt (x,y) een lokaal minimum is volgens de x-as. We moeten dan kijken of f (x-delta, y) en f (x+delta,y) groter dan f (x,y).  We leiden hiervoor van de functie met twéé parameters een functie af die maar één parameter meer heeft, en gebruiken die functie in de bestaande is_lokaal_minimum.

```python
def is_lokaal_minimum_volgens_x(f, x, y):
    def fx(x1):
        return f(x1,y)
    return is_lokaal_minimum(fx, x, 0.1)

import math
def functie2(x,y):
    return math.sin(x)+y

jaja = is_lokaal_minimum_volgens_x(functie2, -math.pi/2, 10)
neenee = is_lokaal_minimum_volgens_x(functie2, 0, 10)
print(f"{jaja} en {neenee}")
```
