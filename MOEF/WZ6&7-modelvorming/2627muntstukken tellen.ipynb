### Muntstukken tellen

Onderstaande figuur toont het aantal muntstukken dat je gelijktijdig moet wegen in functie van de absolute onzekerheid op het totaal aantal muntstukken dat je wilt bekomen.

```python
import matplotlib.pyplot as plt
import numpy as np
#from plotFuncties import *
import ipywidgets

m_1b = 9     # (g) meting van de massa van 1 muntstuk
dm_1b = 1    # (g) absolute onzekerheid op de massa van 1 muntstuk
m_zakje = 5  # (g) meting van de massa van het plastic zakje
dm_zakje = 1 # (g) absolute onzekerheid op de massa van het plastic zakje
m_tot = 1557 # (g) meting van de massa van alle muntstukken
dm_tot = 1   # (g) absolute onzekerheid op de massa van alle muntstukken
dm_a = 1     # (g) absolute onzekerheid op de massa van aantal a muntstukken


def bereken_aantal_te_wegen(dntot):
    return dm_a*((m_tot - m_zakje) / (m_1b ** 2))/(dntot - (1 / m_1b) * dm_tot - (1 / m_1b) * dm_zakje)

def plotAantalMuntstukkenWegen(dntot_min, dntot_max):
    dntot=np.linspace(dntot_min, dntot_max, 250)
    ys = [bereken_aantal_te_wegen(dn) for dn in dntot]
    #of ys = bereken_aantal_te_wegen(dntot)
    plt.plot(dntot, ys, 'b')
    plt.grid(True)
    plt.xlabel("Absolute onzekerheid op aantal muntstukken (-)")
    plt.ylabel("Aantal te wegen muntstukken (-)")
    #plt.xscale("log")
    #plt.yscale("log")
    plt.show()

slider_dntot_min = ipywidgets.FloatSlider(value=0.4, min=0.01, max=0.8, step = 0.05, description ="dntot min")
slider_dntot_max = ipywidgets.FloatSlider(value=0.2, min=1, max=10, step = 0.15, description ="dntot max")
grafiek = ipywidgets.interactive(plotAantalMuntstukkenWegen, dntot_min=slider_dntot_min, dntot_max=slider_dntot_max)
display(grafiek)
```

Bereken de massa muntstukken die je moet wegen om een absolute onzekerheid van 0.4 te bekomen op het berekende aantal muntstukken
in functie van de totale massa muntstukken dat je moet tellen. 

```python
def bereken_aantal_te_wegen2(ntot):
    mtot = ntot * m_1b + m_zakje
    dntot = 0.4
    return dm_a*((mtot - m_zakje) / (m_1b ** 2))/(dntot - (1 / m_1b) * dm_tot - (1 / m_1b) * dm_zakje)

def bereken_massa_te_wegen2(mtot):
    dntot = 0.4
    return dm_a*((mtot - m_zakje) / (m_1b ** 2))/(dntot - (1 / m_1b) * dm_tot - (1 / m_1b) * dm_zakje)*m_1b

def plot_aantal_muntstukken_wegen(mtotMin, mtotMax):
    mtot=np.linspace(mtotMin,mtotMax,250)
    ys = [bereken_massa_te_wegen2(m) for m in mtot]
    # of ys = bereken_massa_te_wegen2(mtot)
    plt.plot(mtot, ys, 'b')
    plt.grid(True)
    plt.xlabel("Massa totaal muntstukken (g)")
    plt.ylabel("Massa te wegen muntstukken (g)")
    plt.title("Massa muntstukken te wegen voor dntot = 0.4 (-)")
    #plt.xscale("log")
    #plt.yscale("log")
    plt.show()






slider_dmtot_min = ipywidgets.FloatSlider(value=m_1b, min=m_1b, max=100, step = 5, description ="mtot Min")
slider_dmtot_max = ipywidgets.FloatSlider(value=2000, min=10, max=5000, step = 10, description ="mtot Max")
grafiek = ipywidgets.interactive(plot_aantal_muntstukken_wegen, mtotMin=slider_dmtot_min, mtotMax=slider_dmtot_max)
display(grafiek)
```

Bereken de massa muntstukken die je moet wegen om een absolute onzekerheid van 0.4 te bekomen op het berekende aantal muntstukken
in functie van de totale massa muntstukken dat je moet tellen wanneer de onzekerheid op het zakje niet wordt meegenomen.

```python
def bereken_aantal_te_wegen3(ntot):
    mtot = ntot * m_1b
    dntot = 0.4
    return dm_a*((mtot) / (m_1b ** 2))/(dntot - (1 / m_1b) * dm_tot)

def bereken_massa_te_wegen3(mtot):
    dntot = 0.4
    return dm_a*((mtot) / (m_1b ** 2))/(dntot - (1 / m_1b) * dm_tot)*m_1b

def plot_aantal_muntstukken_wegen3(mtot_min, mtot_max):
    mtot=np.linspace(mtot_min, mtot_max, 250)
    ys = [bereken_massa_te_wegen3(m) for m in mtot]
    # of ys = bereken_massa_te_wegen3(mtot)
    plt.plot(mtot, ys, 'b')
    plt.grid(True)
    plt.xlabel("Massa totaal muntstukken (g)")
    plt.ylabel("Massa te wegen muntstukken (g)")
    plt.title("Massa muntstukken te wegen voor dntot = 0.4 zonder onzekerheid op het zakje (-)")
    #plt.xscale("log")
    #plt.yscale("log")
    plt.show()

slider_dmtot_min = ipywidgets.FloatSlider(value=m_1b, min=m_1b, max=100, step = 5, description ="mtotMin")
slider_dmtot_max = ipywidgets.FloatSlider(value=2000, min=10, max=5000, step = 10, description ="mtotMax")
grafiek = ipywidgets.interactive(plot_aantal_muntstukken_wegen3, mtot_min=slider_dmtot_min, mtot_max=slider_dmtot_max)
display(grafiek)
```
