### Analyse slinger van Huygens

```python
import matplotlib.pyplot as plt
%matplotlib widget
import numpy as np
import ipywidgets
```

```python
T = 4.00     # (s) meting van de 1 periode van de slinger
dT = 0.01     # (s) absolute onzekerheid op de meting van de 1 periode van de slinger
g = 9.81     # (m/s²) valversnelling
dg = 0.01    # (m/s²) absolute onzekerheid op de valversnelling

def aantal_periodes(dL):
    return 2*g*T*dT/(4*np.pi**2*dL-T**2*dg)

# figuur 1
xs = np.linspace(0.1,1,250)
ys = [aantal_periodes(x) for x in xs]
# of: ys = aantal_periodes(xs)
plt.figure()
plt.plot(xs, ys, color = 'b')
plt.grid(True)
plt.xlabel("absolute onzekerheid op de lengte van het touw (m)")
plt.ylabel("Aantal te meten periodes (-)")
plt.show()


```

```python
def plot_dynamisch(t1):
    n_max = 20
    def bereken_dL_ifv_dT(n1):
        return t1**2*dg/(4*np.pi**2)+g*t1*dT/(2*np.pi**2*n1)
    xs = np.linspace(0.1,n_max,250)
    ys = [bereken_dL_ifv_dT(t) for t in xs]
    #of ys = bereken_dL_ifv_dT(xs)
    plt.figure()
    plt.plot(xs, ys, color = 'b')
    plt.grid(True)
    plt.xlabel("Aantal gemeten periodes (-)")
    plt.ylabel("Absolute onzekerheid op de lengte van het touw (m)")
    plt.ylim(0 , 0.4)
    #plt.yscale("log")
    plt.show()

sliderT = ipywidgets.IntSlider(value=5, min=0.5, max=100, step = 1, description = "periode (s)")
grafiek = ipywidgets.interactive(plot_dynamisch, t1=sliderT)
display(grafiek)
```

```python
T = 4.00     # (s) meting van de 1 periode van de slinger
dT = 0.01     # (s) absolute onzekerheid op de meting van de 1 periode van de slinger
g = 9.81     # (m/s²) valversnelling
dg = 0.01    # (m/s²) absolute onzekerheid op de valversnelling
x_asym= T ** 2 * dg / 4 / np.pi ** 2

 
def plot_aantal_periodes(dL_min, dL_max):
    xs =np.linspace(dL_min, dL_max, 250)
    ys = [aantal_periodes(x) for x in xs]
    # of: ys = aantal_periodes(xs)
    plt.figure()
    plt.plot(xs, ys, color = 'b')
    plt.grid(True)
    plt.xlabel("Gewenste absolute onzekerheid op de lengte (m)")
    plt.ylabel("Aantal periodes te meten (-)")
    #plt.xscale("log")
    #plt.yscale("log")
    plt.show()

slider_dl_min = ipywidgets.FloatSlider(value=0.5, min=x_asym, max=0.8, step = 0.05, description ="dLMin")
slider_dL_max = ipywidgets.FloatSlider(value=5, min=1, max=10, step = 0.20, description ="dLMax")
grafiek = ipywidgets.interactive(plot_aantal_periodes, dL_min=slider_dl_min, dL_max=slider_dL_max)
display(grafiek)
```
